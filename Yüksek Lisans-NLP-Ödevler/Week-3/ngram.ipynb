```python
import nltk
import random
from nltk.util import ngrams
from collections import Counter
```

```python
text = """ahmet ve mehmet beraber sinemaya gitti. mehmet ve ayşe tiyatroya gitti. 
mehmet beraber yemek yedi. mehmet ve ahmet futbol oynadı. mehmet beraber ders çalıştı. 
ahmet ve mehmet sinemaya tekrar gitti. mehmet beraber kitap okudu. mehmet ve ayşe kütüphaneye gitti. 
mehmet beraber yürüyüş yaptı. mehmet ve ayşe çay içti. ahmet ve mehmet dizi izledi."""

# Tokenization işlemi
tokens = nltk.word_tokenize(text.lower())  # Küçük harfe çevir ve kelimelere ayır
print(tokens[:10])  # İlk 10 kelimeyi görelim
```

```python
# Unigram
# Metni temizleyip tokenization yapalım
words = text.lower().replace(".", "").split()

# Kelime frekanslarını hesaplayalım
word_counts = Counter(words)

# Toplam kelime sayısını bulalım
total_words = sum(word_counts.values())

# Unigram olasılıklarını hesaplayalım
word_probs = {word: count / total_words for word, count in word_counts.items()}

# En sık geçen kelimeyi bulalım
most_common_word = max(word_counts, key=word_counts.get)
# Unigram modeline göre rastgele kelime üretelim
def generate_unigram_word():
    words_list = list(word_probs.keys())  # Kelimeleri al
    probs_list = list(word_probs.values())  # Olasılıkları al
    return random.choices(words_list, weights=probs_list, k=1)[0]  # Ağırlıklı rastgele seçim

print(" Kelime Frekansları:", word_counts)
print("\n Unigram Olasılıkları:", word_probs)
print("\n En Sık Geçen Kelime:", most_common_word)
print("\n Unigram Modeli ile Rastgele Kelime:", generate_unigram_word())
```

```python
# Bigramları oluşturalım

bigrams= list(ngrams(tokens, 2))

bigram_freq=Counter(bigrams)

print(bigram_freq.most_common(5))
```

```python
trigrams = list(ngrams(tokens,3))
trigram_freq=Counter(trigrams)
print(trigram_freq.most_common(5))
```

```python
def predict_next_word(word, bigram_freq):
    candidates={}
    for k,v in bigram_freq.items():
        if k[0]==word:
            candidates[k[1]]=v
    if not candidates:
        return "Tahmin bulunamadı"
    return max(candidates, key=candidates.get)
```

```python
print(predict_next_word("mehmet", bigram_freq))  # Kullanıcı "mehmet" derse ne tahmin edilecek?
```

```python
def predict_next_word_trigram(word1, word2, trigram_freq):
    candidates = {k[2]: v for k, v in trigram_freq.items() if k[0] == word1 and k[1] == word2}
    
    if not candidates:
        return "Tahmin bulunamadı."
    
    return max(candidates, key=candidates.get)  # En yüksek frekanslı kelimeyi döndür


print(predict_next_word_trigram("mehmet", "ve", trigram_freq))  # Kullanıcı "mehmet ve" derse?
```
