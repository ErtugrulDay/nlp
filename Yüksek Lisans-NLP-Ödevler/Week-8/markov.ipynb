```python
import re

text = """
Sabah uyanınca pencereyi açtı ve dışarı baktı. Hava griydi ama sessiz bir güzellik taşıyordu.
Sokaktan geçen birkaç kişi hızlı adımlarla yürüyordu. Kahvesini hazırlayıp koltuğa oturdu.
Bugün ne yapacağını bilmiyordu ama içinden bir şeyler üretme isteği geliyordu.
Belki yazı yazar, belki dışarı çıkar, belki sadece düşünürdü. Sessizlik içinde, küçük şeylerin değerini fark etti.
"""

def tokenize(text):
    # Noktalama işaretlerini temizleyip küçük harfe çeviriyoruz
    text = re.sub(r'[^\w\s]', '', text)
    return text.lower().split()

tokens = tokenize(text)
print(tokens[:20])  # İlk 20 kelimeyi görelim
```

```python
from collections import defaultdict
import random

def build_markov_chain(tokens):
    chain = defaultdict(list)
    for i in range(len(tokens) - 1):
        current_word = tokens[i]
        next_word = tokens[i + 1]
        chain[current_word].append(next_word)
    return chain

markov_chain = build_markov_chain(tokens)

# Örnek: "belki" kelimesinden sonra gelenleri görelim
print("belki →", markov_chain["belki"])
```

```python
def generate_sentence(chain, start_word, length=15):
    word = start_word
    sentence = [word]
    
    for _ in range(length - 1):
        next_words = chain.get(word)
        if not next_words:
            break
        word = random.choice(next_words)
        sentence.append(word)
    
    return ' '.join(sentence)

# Örnek bir üretim: "bugün" kelimesiyle başlasın
print(generate_sentence(markov_chain, "bugün"))
```

```python
# Başlangıç kelimesini tokens listesinden rastgele seçerek üretim
random_start = random.choice(tokens)
print(f"Başlangıç kelimesi: {random_start}")
print(generate_sentence(markov_chain, random_start))
```

```python
from collections import defaultdict
# trigram
def build_trigram_chain(tokens):
    chain = defaultdict(list)
    for i in range(len(tokens) - 2):
        key = (tokens[i], tokens[i+1])  # ikili tuple (kelime1, kelime2)
        next_word = tokens[i + 2]
        chain[key].append(next_word)
    return chain

trigram_chain = build_trigram_chain(tokens)

# Örnek bir geçişi görelim
example_key = ('belki', 'dışarı')
print(f"{example_key} → {trigram_chain[example_key]}")
```

```python
def generate_trigram_sentence(chain, start_words, length=15):
    if start_words not in chain:
        raise ValueError("Verilen başlangıç kelimeleri zincirde bulunamadı.")

    word1, word2 = start_words
    sentence = [word1, word2]

    for _ in range(length - 2):
        next_words = chain.get((word1, word2))
        if not next_words:
            break
        next_word = random.choice(next_words)
        sentence.append(next_word)
        word1, word2 = word2, next_word

    return ' '.join(sentence)

# Örnek: Başlangıç olarak ('belki', 'dışarı') verelim
print(generate_trigram_sentence(trigram_chain, ('belki', 'dışarı')))
```

```python
# Rastgele bir başlangıç ikilisi seç
random_start = random.choice(list(trigram_chain.keys()))
print(f"Rastgele başlangıç: {random_start}")

# Üretilen cümleyi yazdır
print(generate_trigram_sentence(trigram_chain, random_start))
```
