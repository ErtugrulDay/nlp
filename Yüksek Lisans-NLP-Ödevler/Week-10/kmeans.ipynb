```python
import os

def klasorden_haberleri_cek(klasor_yolu, max_adet=700):
    haberler = []
    dosyalar = sorted(os.listdir(klasor_yolu), key=lambda x: int(x.replace('.txt', '')))
    for dosya in dosyalar[:max_adet]:
        tam_yol = os.path.join(klasor_yolu, dosya)
        with open(tam_yol, 'r', encoding='utf-8') as f:
            icerik = f.read().strip()
            if icerik:
                haberler.append(icerik)
    return haberler

# Her kategoriden 700 örnek 
spor = klasorden_haberleri_cek("42bin_haber/news/spor", max_adet=700)
teknoloji = klasorden_haberleri_cek("42bin_haber/news/teknoloji", max_adet=700)
siyaset = klasorden_haberleri_cek("42bin_haber/news/siyaset", max_adet=700)

print(len(spor), len(teknoloji), len(siyaset))  # kontrol
```

```python
tum_haberler = spor + teknoloji + siyaset
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(max_features=3000, stop_words=None)
X = vectorizer.fit_transform(tum_haberler)
```

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters=3, random_state=42)
labels = model.fit_predict(X)
```

```python
from collections import defaultdict

kume_dict = defaultdict(list)

for i, label in enumerate(labels):
    kume_dict[label].append(tum_haberler[i])

# Her kümeden ilk 10 haberi yazalım
for kume_no, haberler in kume_dict.items():
    print(f"\n🔹 Küme {kume_no} ({len(haberler)} haber):")
    for haber in haberler[:10]:
        print("-", haber[:120], "...")
```

```python
yorum = ["Beşiktaş yeni sezon hazırlıklarına başladı."]
yorum_vec = vectorizer.transform(yorum)
tahmin = model.predict(yorum_vec)
print("Yorumun ait olduğu küme:", tahmin[0])
```
