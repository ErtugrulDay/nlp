```python
import stanza
 
```

```python
# Türkçe için bir işlemci
nlp = stanza.Pipeline(lang='tr', processors='tokenize,pos,lemma,depparse')
```

```python
# Örnek cümle
metin = "Ali sabah okula hızlıca yürüdü."

# Metni işle
doc = nlp(metin)
```

```python
# Kelime-bağlılık bilgilerini yaz
for sent in doc.sentences:
    for word in sent.words:
        head = "ROOT" if word.head == 0 else sent.words[word.head - 1].text
        print(f"{word.text} --> {head} ({word.deprel})")
```

```python
#  nsubj – özne	Eylemi yapan kişi
# sabah	flat – düzlem/zaman	
# okula	yürüdü	obl – dolaylı tümleç
# hızlıca	yürüdü	advmod – zarf tümleci	
# yürüdü	ROOT	root – cümlenin kökü
#.	yürüdü	punct – noktalama


   
```

```python
metin2 = "Metin hoca'nın takımı bu sene ligi 4.sırada bitirecek."

# NLP analizi
doc = nlp(metin2)

for sent in doc.sentences:
    for word in sent.words:
        head = "ROOT" if word.head == 0 else sent.words[word.head - 1].text
        print(f"{word.text} --> {head} ({word.deprel})")
```
